## **Instance Attributes**
All of the following Instance attributes are initialized in the constructor from the values of the argument. Notice the __ prefix to make is private.
9.	__size is a str that stores the size of this object. This is initialized in the constructor. It is mutated by the property in #8. It is returned in #7. It is used in #6 to calculate the cost of the pizza.
10.	__toppings a list of string that represents the toppings for this object. This is initialized in the constructor. It is mutated in the add() method and is used in #6 to calculate the cost of the pizza.


```python
class Pizza:
  # Class Attributes
  VALID_SIZES = {"small", "medium", "large", "x-large"}
  PRICES =  {"small": 6.49, "medium":8.49, "large":10.49, "x-large":13.49}

  # Constructors
  def __init__(self, size="medium", toppings=None):
        self.__size = None  # Placeholder for proper validation
        self.size = size  # property size for validation
        self.__toppings = toppings if toppings else ["cheese"] # Initialize toppings list with default "cheese"

  def add(self, toppings):
        if not isinstance(toppings, list) or not all(isinstance(t, str) for t in toppings):
            raise ValueError("Toppings must be a list of strings") # validating the input
        self.__toppings.extend(toppings) # adding toppings to pizza

  # Instance Properties and Attributes
  @property
  def price(self):
    return self.PRICES[self.__size] + (0.50 * len(self.__toppings)) # base price + 0.5 per topping

  @property
  def size(self): # getter method for pizza size
    return self.__size

  @size.setter
  def size(self, new_size): # setter method for pizza size
    if new_size not in self.VALID_SIZES:
      raise ValueError(f"ERROR: {new_size} is not a valid size for a pizza") # validating size input
    self.__size = new_size # update the size if valid

  def __str__(self):
    return f"{self.__size} pizza with {self.__toppings} for ${self.price:.2f}" # string format to display pizza details
```

```python
# Test Harness
print(f'Creating a default pizza')
p = Pizza()
print(p)

toppings = 'cheese olive'.split()
print(f'\nAdding topping: {toppings}')
p.add(toppings=toppings)
print(p)

print(f'\nCreating a new pizza')
p = Pizza('large', 'cheese pepper'.split())
print(p)

toppings = ['pineapple', 'mushroom']
print(f'\nAdding topping: {toppings}')
p.add(toppings)
print(p)

size = 'x-large'
p.size = size
print(f'\nChanging order size to {size}')
print(p)

size = 'gigantic'
print(f'\nChanging order size to {size}')
try:
  p.size = size
except ValueError as err:
  print(err)
```
